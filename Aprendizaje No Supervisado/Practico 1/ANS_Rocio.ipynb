```python
df["primary_position"] = (
    df["player_positions"]
    .str.split(",")
    .str[0]
    .str.strip()
)

df["primary_position"].value_counts()
```

```python
posiciones = df["primary_position"].value_counts()

plt.figure(figsize=(12, 6))

sns.barplot(
    x=posiciones.index,
    y=posiciones.values
)

plt.title("Distribución de jugadores por posición principal")
plt.xlabel("Posición")
plt.ylabel("Cantidad de jugadores")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
def agrupar_posicion(position):

    forwards = [
        "ST", "CF", "LW", "RW"
    ]

    midfielders = [
        "CAM", "CM", "CDM", "LM", "RM"
    ]

    defenders = [
        "CB", "LB", "RB", "LWB", "RWB"
    ]

    goalkeepers = [
        "GK"
    ]

    if position in forwards:
        return "Forward"

    elif position in midfielders:
        return "Midfielder"

    elif position in defenders:
        return "Defender"

    elif position in goalkeepers:
        return "GK"

    else:
        return "Other"
```

```python
df["Position2"] = (
    df["primary_position"]
    .apply(agrupar_posicion)
)

df["Position2"].value_counts()
```

```python
posiciones_generales = df["Position2"].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=posiciones_generales.index,
    y=posiciones_generales.values
)

plt.title("Distribución de jugadores por grupo de posición")
plt.xlabel("Grupo de posición")
plt.ylabel("Cantidad de jugadores")
plt.tight_layout()
plt.show()
```
