```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Asegurar que los valores numéricos lo sean
# Lista de columnas que por naturaleza de este conjunto de datos deberían ser numéricas
cols_to_numeric = ['Price', 'Rooms', 'Distance', 'Postcode', 'Bathroom',
                   'Car', 'Landsize', 'BuildingArea', 'YearBuilt', 'Lattitude', 'Longtitude']

# Forzamos la conversión a numérico. Los valores ilegibles se convertirán en NaN (errors='coerce')
for col in cols_to_numeric:
    if col in melb_df.columns:
        melb_df[col] = pd.to_numeric(melb_df[col], errors='coerce')

# Filtramos solo las columnas numéricas para evitar errores en la matriz de correlación
melb_numeric = melb_df.select_dtypes(include=['number', 'float64', 'int64']).copy()

# Eliminamos filas donde el precio es nulo, ya que no nos sirven para calcular correlación
melb_numeric = melb_numeric.dropna(subset=['Price'])
print ("\n melb_numeric datos: \n")
melb_numeric.info()


# 2. Descartar el 1% de valores mayores y el 1% inferior de Price
q01 = melb_numeric['Price'].quantile(0.01)
q99 = melb_numeric['Price'].quantile(0.99)

# Filtramos el DataFrame conservando el 98% central de los datos
melb_filtered = melb_numeric[(melb_numeric['Price'] >= q01) & (melb_numeric['Price'] <= q99)]

print(f"Filas originales con Precio válido: {len(melb_numeric)}")
print(f"Filas conservadas tras el recorte del 1% en ambos extremos: {len(melb_filtered)}")
print ("\n melb_filtered datos: \n")
melb_filtered.info()

# 3. Calcular los coeficientes de correlación
correlaciones = melb_filtered.corr()

# Aislamos la columna 'Price', quitamos la auto-correlación (Price vs Price = 1) y ordenamos
corr_con_precio = correlaciones['Price'].drop('Price').sort_values(ascending=False)

# 4. Generar el gráfico de coeficientes
plt.figure(figsize=(12, 7))

# Asignamos colores: un color para correlación positiva y otro para negativa
colores = ['#2ca02c' if valor > 0 else '#d62728' for valor in corr_con_precio.values]

sns.barplot(x=corr_con_precio.values, y=corr_con_precio.index, palette=colores)

# Configuración estética del gráfico
plt.title('Correlación de las variables con el Precio de la Propiedad', fontsize=16, fontweight='bold')
plt.xlabel('Coeficiente de Correlación de Pearson', fontsize=12)
plt.ylabel('Variables Numéricas', fontsize=12)

# Línea central en cero
plt.axvline(x=0, color='black', linestyle='-', linewidth=1)

# Líneas guía de umbral (ej. 0.2 y -0.2) para ayudar a decidir qué descartar
umbral_visual = 0.2
plt.axvline(x=umbral_visual, color='gray', linestyle='--', linewidth=1.5, alpha=0.7)
plt.axvline(x=-umbral_visual, color='gray', linestyle='--', linewidth=1.5, alpha=0.7)

plt.tight_layout()
plt.show()

# 5. Selección final de columnas
# Seleccionamos automáticamente aquellas que superan el umbral (ya sea positivo o negativo)
columnas_seleccionadas = corr_con_precio[abs(corr_con_precio) >= umbral_visual].index.tolist()

# Agregamos 'Price' de vuelta a la lista, ya que es nuestro target
columnas_seleccionadas.append('Price')

print(f"\nColumnas seleccionadas con correlación absoluta mayor o igual a {umbral_visual}:")
print(columnas_seleccionadas)

# Si deseas aplicar esta selección al DataFrame original:
# melb_df_final = melb_df[columnas_seleccionadas]
```

```python
# Crear el motor de SQLite (archivo local 'melb_housing.db')
engine = create_engine('sqlite:///melb_housing.db', echo=False)

# Ingestar los datos en tablas
melb_df.to_sql('melb_data', con=engine, if_exists='replace', index=False)
df_airbnb.to_sql('airbnb_data', con=engine, if_exists='replace', index=False)

print("Tablas 'melb_data' y 'airbnb_data' creadas exitosamente.")
```

```python
melb_df_2= melb_df.copy()

# 1. Convertir Date a datetime
melb_df_2['Date'] = pd.to_datetime(melb_df['Date'], dayfirst=True)

# 2. Asegurar que Price sea float
melb_df_2['Price'] = pd.to_numeric(melb_df['Price'], errors='coerce')

# 3. Limpiar Postcode (clave para el JOIN posterior)
melb_df_2['Postcode'] = melb_df['Postcode'].astype(float)
df_airbnb['zipcode'] = df_airbnb['zipcode'].astype(float)

print("Tipos de datos de Melbourne:\n", melb_df_2.dtypes[['Date', 'Price', 'Postcode']])

print("Tipos de datos de Airbnb:\n", df_airbnb.dtypes[['zipcode', 'airbnb_price_mean', 'airbnb_record_count']])
```

```python
# A. Cantidad de registros totales por Regionname
query_a = "SELECT Regionname, COUNT(*) as total FROM melb_data GROUP BY Regionname"
res_a = pd.read_sql(query_a, engine)

# B. Cantidad de registros totales por Suburb y Regionname
query_b = "SELECT Suburb, Regionname, COUNT(*) as total FROM melb_data GROUP BY Suburb, Regionname"
res_b = pd.read_sql(query_b, engine)

# C. Propiedades por Regionname con más de 2 habitaciones
query_c = """
SELECT Regionname, COUNT(*) as total
FROM melb_data
WHERE Rooms > 2
GROUP BY Regionname
"""
res_c = pd.read_sql(query_c, engine)

# D. Precio promedio según tipo (Type) y Regionname
query_d = """
SELECT Type, Regionname, AVG(Price) as avg_price
FROM melb_data
GROUP BY Type, Regionname
"""
res_d = pd.read_sql(query_d, engine)

# E. Top 5 barrios con propiedades más caras en promedio
query_e = """
SELECT Suburb, AVG(Price) as avg_price
FROM melb_data
GROUP BY Suburb
ORDER BY avg_price DESC
LIMIT 5
"""
res_e = pd.read_sql(query_e, engine)

display(res_e)
```

```python
# Combinar los datasets mediante JOIN
query_join = """
SELECT m.*, a.airbnb_price_mean, a.airbnb_record_count
FROM melb_data m
LEFT JOIN airbnb_data a ON m.Postcode = a.zipcode
"""
df_combined = pd.read_sql(query_join, engine)

# --- CELDA DE VALIDACIÓN POSTERIOR ---
def validate_data(df_orig, df_res):
    # 1. Verificar que el número de filas de la tabla principal no cambió (LEFT JOIN)
    assert len(df_orig) == len(df_res), f"Error: Se perdieron o duplicaron filas. Original: {len(df_orig)}, Result: {len(df_res)}"

    # 2. Verificar nulos inesperados en columnas clave
    # Es normal tener algunos nulos si no había coincidencia en Airbnb,
    # pero no debería haber nulos en los datos originales de Melbourne.
    assert df_res['Price'].notnull().all(), "Error: Aparecieron nulos en la columna Price"

    # 3. Rangos de variables razonables
    # El precio promedio de Airbnb no debería ser negativo ni exageradamente alto para Melbourne
    if df_res['airbnb_price_mean'].notnull().any():
        mean_price = df_res['airbnb_price_mean'].dropna()
        assert (mean_price >= 0).all(), "Error: Hay precios de Airbnb negativos"
        assert (mean_price < 5000).all(), "Error: Hay precios de Airbnb fuera de rango razonable (>5000)"

    print("Validación de calidad completada: Validez, Completitud e Integridad OK.")

validate_data(melb_df_2, df_combined)

```

```python
import anthropic
import json

client = anthropic.Anthropic()

# Valores únicos con posibles inconsistencias
council_values = melb_df['CouncilArea'].dropna().unique().tolist()

message = client.messages.create(
    model="claude-opus-4-7",
    max_tokens=1024,
    messages=[{
        "role": "user",
        "content": f"""Estos son los valores únicos de la columna CouncilArea en un dataset de propiedades de Melbourne:
{council_values}

Identificá: (1) duplicados con distinta capitalización o spelling,
 (2) valores que parecen errores, (3) valores que podrían agruparse.
Respondé en JSON con la estructura: {{"estandarizado": {{"valor_original": "valor_correcto"}}}}"""
    }]
)

mapping = json.loads(message.content[0].text)
melb_df['CouncilArea_clean'] = melb_df['CouncilArea'].map(
    mapping.get('estandarizado', {})
).fillna(melb_df['CouncilArea'])
```
